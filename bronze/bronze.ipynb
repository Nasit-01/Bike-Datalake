Reading csv from row source


```python
df = (
    spark.read.option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/dev_project/bronze/raw_sources/cust_info.csv")
)
df.display()
```

Write into bronze layer

```python
df.write.mode("overwrite").saveAsTable("dev_project.bronze.crm_cust_info")
```
